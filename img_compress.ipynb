```python
import fitz  # PyMuPDF
from PIL import Image
import io

def compress_pdf_images(input_path, output_path, image_quality=50):
    # Open the PDF file
    doc = fitz.open(input_path)

    for page_num in range(len(doc)):
        page = doc[page_num]
        images = page.get_images(full=True)

        for img_info in images:
            xref = img_info[0]  # The reference number for the image
            base_image = doc.extract_image(xref)
            image_bytes = base_image["image"]
            image_ext = base_image["ext"]

            # Compress the image using Pillow
            image = Image.open(io.BytesIO(image_bytes))
            img_buffer = io.BytesIO()
            image = image.convert("RGB")  # Convert to RGB for compatibility
            image.save(img_buffer, format="JPEG", quality=image_quality)
            
            # Insert the compressed image back into the PDF
            img_buffer.seek(0)
            rect = fitz.Rect(0, 0, image.width, image.height)  # Placeholder for image rect
            page.insert_image(rect, stream=img_buffer)

    # Save the compressed PDF
    doc.save(output_path, garbage=4, deflate=True)
    doc.close()

# Example usage
input_pdf = "assets/pdf/cde_sem_1/2_volcano_eruption_in_ulawun_analysis.pdf"
output_pdf = "assets/pdf/cde_sem_1/2_volcano_eruption_in_ulawun_analysis_compressed123cs.pdf"

compress_pdf_images(input_pdf, output_pdf)
```
